Напишите функцию `def normalize_items(df)`, которая приводит набор к единой схеме: преобразует id, value, updated_at, active, нормализует текстовые категории и удаляет полные дубликаты.

````python
items_csv.columns
````

````python
def normalize_items(df):
    df = df.copy()
    df['id'] = pd.to_numeric(df['id'], errors='coerce').astype('Int64')
    
    df['value'] = df['value'].astype(str).str.replace(',', '.')
    df['value'] = pd.to_numeric(df['value'], errors='coerce')
    
    df['updated_at'] = pd.to_datetime(
        df['updated_at'], format='mixed', errors='coerce'
        )
    
    clean_active = (
        df['active'].astype(str).str.strip().str.lower().str.replace('.0', '', regex=False)
    )
    df['active'] = clean_active.map(
        {'true': True, 'yes': True, '1': True, 'да': True,
        'false': False, 'no': False, '0': False, 'нет': False}
        ).fillna(False)

    df['name'] = df['name'].str.strip().str.lower()
    df['category'] = df['category'].str.strip().str.lower()

    df = df.drop_duplicates()
    df = df.dropna(subset=['id'])

    return df
````

````python
csv_normalized = normalize_items(items_csv)
json_normalized = normalize_items(items_json)
xml_normalized = normalize_items(items_xml)
````

Объедините три файла в один DataFrame.

````python
df = pd.concat([csv_normalized, json_normalized, xml_normalized], ignore_index=True)
df.head()
````

#### Этап C. Обработка ошибок.

Реализуйте обработку как минимум трех ситуаций:

- несуществующий id;
- ошибка соединения;
- превышение timeout.

Программа не должна завершаться необработанным traceback. Ошибка должна попасть в лог вместе с URL и ее типом. Requests предоставляет единое семейство исключений RequestException, включая ConnectionError, Timeout и HTTP-ошибки.

````python
def send_request(method, url, **kwargs):
    start_time = time.time()

    try:
        response = requests.request(
            method,
            url,
            **kwargs
        )
        response.raise_for_status()

        result = {
            'method': response.request.method,
            'url': url,
            'params/body': kwargs.get('json') or kwargs.get('params'),
            'status_code': response.status_code,
            'response_time': time.time() - start_time,
            'response_data': response.json()
        }

        results.append(result)
        return result

    except requests.exceptions.RequestException as e:
        result = {
            'method': method,
            'url': url,
            'params/body': kwargs.get('json') or kwargs.get('params'),
            'status_code': response.status_code if 'response' in locals() else None,
            'response_time': time.time() - start_time,
            'error': type(e).__name__,
            'response_data': str(e)
        }

        results.append(result)
        return result
````

````python
send_request(
    'GET',
    base_url,
    timeout=0.0000001
)
````

#### Этап E. Сравнение способов получения данных.

````python
comparison = []

# CSV / JSON / XML
start_time = time.time()
formats_data = pd.concat(
    [csv_normalized, json_normalized, xml_normalized],
    ignore_index=True
)

formats_data = formats_data.sort_values(
    by=['id', 'updated_at'],
    ascending=True,
    na_position='first'
)

formats_data = formats_data.drop_duplicates(
    subset='id',
    keep='last'
)

format_time = time.time() - start_time

comparison.append({
    'Способ': 'CSV/XML/JSON',
    'Количество записей': len(formats_data),
    'Уникальных id': formats_data['id'].nunique(),
    'Время': round(format_time, 4),
    'Есть ошибки': 'Нет'
})


# REST API
start_time = time.time()
api_error = False

try:
    response = requests.get(
        base_url,
        timeout=5
    )

    response.raise_for_status()
    api_data = response.json()

except requests.exceptions.RequestException as e:
    api_data = []
    api_error = True

api_time = time.time() - start_time

comparison.append({
    'Способ': 'REST API',
    'Количество записей': len(api_data),
    'Уникальных id': len(set(item['id'] for item in api_data)),
    'Время': round(api_time, 4),
    'Есть ошибки': 'Да' if api_error else 'Нет'
})


# Selenium
start_time = time.time()
selenium_error = False
try:
    from selenium import webdriver
    from selenium.webdriver.common.by import By
    from selenium.webdriver.support import expected_conditions as EC
    from selenium.webdriver.support.ui import WebDriverWait, Select

    driver = webdriver.Chrome()
    driver.get('http://localhost:8000/dynamic')
    button = WebDriverWait(driver, 10).until(
        EC.element_to_be_clickable((By.TAG_NAME, 'button'))
    )
    button.click()
    WebDriverWait(driver, 10).until(
        EC.presence_of_element_located(
            (By.CSS_SELECTOR, '#items tr')
        )
    )
    rows = driver.find_elements(
        By.CSS_SELECTOR,
        '#items tr'
    )

    selenium_data = []
    for row in rows:
        cells = row.find_elements(By.TAG_NAME, 'td')

        if len(cells) == 5:
            selenium_data.append({
                'id': int(cells[0].text),
                'name': cells[1].text,
                'category': cells[2].text,
                'value': float(cells[3].text.replace(',', '.')),
                'active': cells[4].text.strip().lower() == 'true'
            })

    selenium_data = pd.DataFrame(selenium_data)

except Exception as e:
    selenium_data = pd.DataFrame()
    selenium_error = True
    print('Ошибка Selenium:', type(e).__name__, e)

finally:
    if 'driver' in locals():
        driver.quit()

selenium_time = time.time() - start_time

comparison.append({
    'Способ': 'Selenium',
    'Количество записей': len(selenium_data),
    'Уникальных id': (
        selenium_data['id'].nunique()
        if not selenium_data.empty else 0
    ),
    'Время': round(selenium_time, 4),
    'Есть ошибки': 'Да' if selenium_error else 'Нет'
})

comparison_df = pd.DataFrame(comparison)
comparison_df
````
